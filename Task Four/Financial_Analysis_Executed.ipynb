# 10-K Financial Data Analysis: Microsoft, Tesla & Apple

## Objective
Analyze key financial data extracted from SEC EDGAR 10-K filings for **Microsoft**, **Tesla**, and **Apple** over the last three fiscal years (FY2022–FY2024). This analysis will identify trends and insights that could inform the development of an AI-powered financial chatbot.

## Methodology
1. **Data Extraction** — Financial figures manually extracted from 10-K filings on SEC EDGAR
2. **Data Loading** — CSV loaded into pandas DataFrame
3. **Trend Analysis** — Year-over-year growth rates calculated for all metrics
4. **Comparative Analysis** — Cross-company comparisons and insights
5. **Conclusions** — Key findings for AI chatbot development

### Data Sources
- [Microsoft 10-K (SEC EDGAR)](https://www.sec.gov/cgi-bin/browse-edgar?action=getcompany&CIK=0000789019&type=10-K)
- [Tesla 10-K (SEC EDGAR)](https://www.sec.gov/cgi-bin/browse-edgar?action=getcompany&CIK=0001318605&type=10-K)
- [Apple 10-K (SEC EDGAR)](https://www.sec.gov/cgi-bin/browse-edgar?action=getcompany&CIK=0000320193&type=10-K)

---
## Step 1: Import Libraries and Load Data

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

# Load extracted financial data
df = pd.read_csv('financial_data.csv')
print(f'Dataset Shape: {df.shape}')
print(f'\nFinancial Data (All figures in USD):')
df
```

---
## Step 2: Data Formatting — Convert to Billions for Readability

```python
# Create a readable version in billions
df_display = df.copy()
numeric_cols = ['Total Revenue', 'Net Income', 'Total Assets', 'Total Liabilities', 'Cash Flow from Operating Activities']

df_display['Revenue ($B)'] = (df['Total Revenue'] / 1e9).round(2)
df_display['Net Income ($B)'] = (df['Net Income'] / 1e9).round(2)
df_display['Total Assets ($B)'] = (df['Total Assets'] / 1e9).round(2)
df_display['Total Liabilities ($B)'] = (df['Total Liabilities'] / 1e9).round(2)
df_display['Operating Cash Flow ($B)'] = (df['Cash Flow from Operating Activities'] / 1e9).round(2)

display_cols = ['Company', 'Fiscal Year', 'Revenue ($B)', 'Net Income ($B)', 'Total Assets ($B)', 'Total Liabilities ($B)', 'Operating Cash Flow ($B)']
print('\nFinancial Data (in Billions USD):')
print(df_display[display_cols].to_string())
```

---
## Step 3: Year-over-Year Growth Analysis

```python
# Calculate Year-over-Year growth rates
df['Revenue Growth (%)'] = df.groupby('Company')['Total Revenue'].pct_change() * 100
df['Net Income Growth (%)'] = df.groupby('Company')['Net Income'].pct_change() * 100
df['Asset Growth (%)'] = df.groupby('Company')['Total Assets'].pct_change() * 100
df['Liabilities Growth (%)'] = df.groupby('Company')['Total Liabilities'].pct_change() * 100
df['Operating CF Growth (%)'] = df.groupby('Company')['Cash Flow from Operating Activities'].pct_change() * 100

growth_cols = ['Company', 'Fiscal Year', 'Revenue Growth (%)', 'Net Income Growth (%)', 'Asset Growth (%)', 'Liabilities Growth (%)', 'Operating CF Growth (%)']
print('\n=== Year-over-Year Growth Rates (%) ===')
print(df[growth_cols].dropna().round(2).to_string())
```

---
## Step 4: Profitability and Efficiency Ratios

```python
# Calculate key financial ratios
df['Profit Margin (%)'] = (df['Net Income'] / df['Total Revenue'] * 100).round(2)
df['Debt-to-Asset Ratio'] = (df['Total Liabilities'] / df['Total Assets']).round(2)
df['OCF-to-Revenue (%)'] = (df['Cash Flow from Operating Activities'] / df['Total Revenue'] * 100).round(2)

ratio_cols = ['Company', 'Fiscal Year', 'Profit Margin (%)', 'Debt-to-Asset Ratio', 'OCF-to-Revenue (%)']
print('\n=== Key Financial Ratios ===')
print(df[ratio_cols].to_string())
```

---
## Step 5: Comparative Summary by Company

```python
# Average metrics by company
summary = df.groupby('Company').agg({
    'Total Revenue': 'mean',
    'Net Income': 'mean',
    'Profit Margin (%)': 'mean',
    'Debt-to-Asset Ratio': 'mean',
    'Cash Flow from Operating Activities': 'mean'
}).round(2)

summary.columns = ['Revenue ($B)', 'Net Income ($B)', 'Profit Margin (%)', 'Debt-to-Asset', 'OCF ($B)']
summary['Revenue ($B)'] = (summary['Revenue ($B)'] / 1e9).round(2)
summary['Net Income ($B)'] = (summary['Net Income ($B)'] / 1e9).round(2)
summary['OCF ($B)'] = (summary['OCF ($B)'] / 1e9).round(2)

print('\n=== Average Metrics by Company (FY2022-FY2024) ===')
print(summary.to_string())

# Revenue ranking for latest year
latest = df[df['Fiscal Year'] == 2024][['Company', 'Total Revenue']].copy()
latest['FY2024 Revenue ($B)'] = (latest['Total Revenue'] / 1e9).round(2)
print('\n=== FY2024 Revenue Ranking ===')
print(latest[['Company', 'FY2024 Revenue ($B)']].sort_values('FY2024 Revenue ($B)', ascending=False).to_string())
```

---
## Step 6: Trend Analysis — Revenue Over 3 Years

```python
# Revenue pivot
rev_pivot = df.pivot_table(index='Company', columns='Fiscal Year', values='Total Revenue')
rev_pivot = (rev_pivot / 1e9).round(2)
rev_pivot['3-Year Change (%)'] = ((rev_pivot[2024] - rev_pivot[2022]) / rev_pivot[2022] * 100).round(2)
print('\n=== Revenue Trends ($B) ===')
print(rev_pivot.to_string())

# Net Income pivot
ni_pivot = df.pivot_table(index='Company', columns='Fiscal Year', values='Net Income')
ni_pivot = (ni_pivot / 1e9).round(2)
ni_pivot['3-Year Change (%)'] = ((ni_pivot[2024] - ni_pivot[2022]) / ni_pivot[2022] * 100).round(2)
print('\n=== Net Income Trends ($B) ===')
print(ni_pivot.to_string())

# Cash Flow pivot
cf_pivot = df.pivot_table(index='Company', columns='Fiscal Year', values='Cash Flow from Operating Activities')
cf_pivot = (cf_pivot / 1e9).round(2)
cf_pivot['3-Year Change (%)'] = ((cf_pivot[2024] - cf_pivot[2022]) / cf_pivot[2022] * 100).round(2)
print('\n=== Operating Cash Flow Trends ($B) ===')
print(cf_pivot.to_string())
```

---
## Step 7: Key Findings and Insights

### Microsoft — Strong and Accelerating Growth
- **Revenue** grew 23.6% over 3 years, accelerating from 6.9% (FY23) to 15.7% (FY24)
- **Net Income** surged 21.8% in FY2024 after being flat in FY2023, driven by AI/cloud growth
- **Profit margin** remains the highest of all three companies at ~36%
- **Operating cash flow** jumped 35% in FY2024 — strongest cash generation growth
- **Insight**: Microsoft's AI investments (Copilot, Azure OpenAI) are clearly paying off

### Tesla — Revenue Growth Stalling, Profitability Declining
- **Revenue** growth collapsed from 18.8% (FY23) to just 0.95% (FY24)
- **Net Income** dropped 52.7% in FY2024 — the sharpest decline of all three companies
- **Profit margin** halved from 15.4% to 7.3% due to aggressive price cuts
- **Operating cash flow** declined for two consecutive years (-10% then -13%)
- **Insight**: Price war strategy is hurting margins significantly; growth model under pressure

### Apple — Stable Revenue, Gradual Margin Compression
- **Revenue** dipped 2.8% in FY2023 then recovered modestly (+2%) in FY2024
- **Net Income** has declined for two consecutive years (-2.8%, -3.4%)
- **Profit margin** stable around 24-25% but trending slightly downward
- **Debt-to-Asset ratio** is highest at 0.84 — Apple uses significant leverage
- **Insight**: Hardware saturation visible; Services segment is critical for future growth
